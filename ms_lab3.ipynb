Имеются результаты измерения некоторой величины:

```python
# @title
import numpy as np
from IPython.display import display, Markdown
from scipy import stats   # для t_gamma (распределение Стьюдента) и t из Ф(t)

# ─── Данные вариантов 1–8 (вариант 0 — пример из методички, n = 10) ───
variant_data = {
    0: [2.2, 4.5, 3.8, 5.6, 4.6, 3.0, 2.3, 3.5, 4.4, 3.2],       # пример
    1: [20.58, 22.80, 25.40, 26.08, 23.25, 21.42, 23.10, 24.09, 24.02, 26.12, 22.84, 25.10],
    2: [43.4, 43.6, 44.3, 44.5, 44.6, 44.5, 44.7, 44.7, 44.6, 44.8, 45.1, 45.2, 45.3],
    3: [31.85, 31.36, 30.32, 30.90, 31.70, 32.40, 31.60, 31.12, 30.98, 31.02, 31.05, 31.00],
    4: [4.8, 6.2, 6.0, 5.9, 4.9, 6.0, 5.6, 6.1, 5.9, 5.8, 5.7, 5.1, 5.5, 6.2, 5.4],
    5: [20.49, 22.67, 23.40, 25.05, 22.15, 20.42, 24.35, 21.20, 23.05, 24.02],
    6: [30.8, 29.9, 30.3, 29.9, 30.4, 30.2, 27.5, 31.1, 29.2, 29.7, 30.8, 30.3, 31.8, 30.7],
    7: [50.52, 52.24, 51.82, 50.46, 51.88, 52.01, 53.00, 51.12, 51.08, 51.92, 52.08, 51.78],
    8: [26.2, 27.5, 26.0, 32.2, 35.0, 30.2, 29.7, 26.8, 30.0, 30.8, 32.0, 28.8, 35.0, 30.8],
}

variant = 2  # выберите номер своего варианта (1--8; 0 --- пример)

x = np.array(variant_data[variant])
n = len(x)

xstr = ""
for item in x:
  xstr = xstr + str(item) + fr""" \quad """

s1=fr"""
Вариант ${variant}$. Значения $x_i$:

${{{xstr}}}$
"""

display(Markdown(s1))
```

*[Выберите свой вариант в коде]*

Будем рассматривать эти результаты как $n$ реализаций случайной величины $X$. Далее будем предполагать, что $X$ имеет нормальное распределение.

Требуется:

1) найти точечные несмещенные оценки математического ожидания $a$ и среднего квадратического отклонения $\sigma$;

2) найти доверительный интервал, покрывающий математическое ожидание с заданной доверительной вероятностью $\gamma = 0.95;$

3) найти погрешность, с которой среднее арифметическое оценивает математическое ожидание $a$ случайной величины $X$, если доверительная вероятность $\gamma = 0.99$; $\gamma = 0.999$;

4) найти минимальный объем выборки, чтобы с доверительной вероятностью $\gamma = 0{,}95$ можно было утверждать, что, принимая среднее арифметическое за математическое ожидание случайной величины $X$, допускаем погрешность $\varepsilon = \dfrac{\sigma}{3}$. *[В нулевом варианте $\varepsilon = \dfrac{\sigma}{2}$]*

#### Решение

Найдем несмещенные оценки для математического ожидания и среднего квадратического отклонения: $\bar{x}_B = \frac{1}{n}\sum_{i=1}^{n} x_i$, $S = \sqrt{\frac{1}{n-1}\sum_{i=1}^{n} \left(x_i - \bar{x}_B\right)^2}.$

Для этого заполним вспомогательную таблицу 1.


```python
# @title
xb = round(float(x.mean()), 2)                # выборочное среднее


dev = np.round(x - xb, 2)                     # отклонения x_i - x_B
dev_sq = np.round(dev ** 2, 4)                # их квадраты


col_i = np.arange(1, n + 1)

col_xi = [str(xitem) for xitem in x]



# Последняя строка
total_x = round(x.sum(), 1)
total_dev = round(dev.sum(), 2)
total_sq = round(dev_sq.sum(), 3)




# Заголовки столбцов (первая колонка — названия строк)
headers = ["$i$", "$x_i$", fr"$x_i - \overline x_B$", fr"$ (x_i - \overline x_B)^2 $"]







# Начало таблицы
table = "| " + " | ".join(headers) + " |\n"
table += "|" + "|".join(["---"] * len(headers)) + "|\n"

# Перебираем строки данных и добавляем в таблицу
for row in col_i:

    rr = []
    rr.append(str(row))
    rr.append(str(col_xi[row-1]))
    rr.append(str(dev[row-1]))
    rr.append(str(dev_sq[row-1]))

    table += "| " + " | ".join(rr) + " |\n"

rr = []
rr.append("Сумма")
rr.append(str(total_x))
rr.append(str(total_dev))
rr.append(str(total_sq))
table += "| " + " | ".join(rr) + " |\n"
# Красивый вывод
s1 = fr"""
Таблица 1 $-$ Вспомогательные вычисления для $\bar{{x}}_B$ и $S$

"""


display(Markdown(table))
display(Markdown(s1))


sum_sq = round(dev_sq.sum(), 3)     # сумма квадратов отклонений
S2 = sum_sq / (n - 1)               # исправленная дисперсия
S = round(np.sqrt(S2), 3)           # исправленное среднее квадратическое отклонение



s2 = fr"""
Заполнив второй столбец таблицы, найдём $\bar x_B = \frac{{{total_x}}}{{{n}}} \approx \textbf{{{xb}}}$. По последнему столбцу имеем
$S = \sqrt {{ \frac{1}{{{n-1}}} \cdot {total_sq} }} = \sqrt{{ {round(S2, 3)} }} \approx \textbf{{{S}}}$

Таким образом, получили точечные несмещенные оценки математического ожидания $\bar{{x}}_B$ и среднего квадратического отклонения $S$

"""

display(Markdown(s2))
```

Найдем доверительный интервал, покрывающий математическое ожидание с надежностью $\gamma = 0{,}95$. Поскольку $\sigma$ неизвестно, то доверительный интервал имеет вид:

$$ \left( \bar{x}_B -  \frac{t_\gamma \cdot S}{\sqrt{n}};\, \bar{x}_B + \frac{t_\gamma \cdot S}{\sqrt{n}} \right) $$

```python
# @title
import math



# compute t_gamma
gamma = 0.95
t_gamma = round(stats.t.ppf(1 - (1 - gamma)/2, df=n-1), 2)  # t для доверительного интервала при неизвестном sigma
eps = round(t_gamma * S / math.sqrt(n), 2)
left = round(xb - eps, 2)
right = round(xb + eps, 2)

s = fr"""
$t_{{\gamma}}$ найдём по таблице (приложение 4) по $\gamma = 0.95$, $n = {{{ n }}}$: $t_{{\gamma}} = \textbf{{{t_gamma}}}$

Найдем точность оценки: $\varepsilon =  \frac{{t_\gamma \cdot S}}{{\sqrt{{n}}}} =  \cdot \frac{{ {t_gamma} \cdot { S}}}{{\sqrt{{{n}}}}} = \textbf{{{eps}}}$.

Тогда доверительный интервал математического ожидания имеет вид: $({xb} - {eps};\, {xb} + {eps})$ или $({left};\, {right})$, то есть ${left} < a < {right}$.

Смысл полученного результата в том, что если будет проведено достаточно большое число серий, состоящих из $\textbf{{{n}}}$ измерений данной величины, то в {int(gamma*100)}% случаев из них найденный доверительный интервал накроет математическое ожидание.
"""
display(Markdown(s))
```

```python
# @title
gammas = [0.99, 0.999]
t_values = [round(stats.t.ppf(1 - (1 - g)/2, df=n-1), 2) for g in gammas]
eps_values = [round(t * S / math.sqrt(n), 2) for t in t_values]

parts = []

parts.append(fr"""
Выполняем третий пункт задания
""")
for g, t, eps in zip(gammas, t_values, eps_values):
    parts.append(fr"Если $\gamma = {g}$, $n = {n}$, то $t_\gamma = \textbf{{{t}}}$. Точность оценки при этом будет: "
                 fr"$\varepsilon =  \frac{{t_\gamma \cdot S}}{{\sqrt{{n}}}} =   \frac{{ {t} \cdot {S} }}{{\sqrt{{{n}}}}} = \textbf{{{eps}}}$.")
s = "\n\n".join(parts) + "\n\nПри увеличении надежности точность оценки снижается."
display(Markdown(s))
```

Минимальный объем выборки найдем по формуле:
$ n = \frac{t^2 \sigma^2}{\varepsilon^2}. $
Значение $t$ находим из соотношения $2\Phi(t) = \gamma = 0.95$, $\Phi(t) = 0.475$. По таблице значений функции $\Phi(t)$ (приложение 2) получаем $t = 1.96$.

```python
# @title
gamma = 0.95

denom = 3

if (variant == 0):
  denom = 2

t = round(stats.norm.ppf(1 - (1 - gamma)/2), 2)  # из нормального распределения, т.к. σ известно
n_min = math.ceil((t**2 * denom**2))  # поскольку ε = σ/2, то n = t^2 * σ^2 / (σ/2)^2 = t^2 * 4
# Но чтобы показать формулу, можно вычислить через выражение с сигмой, но оно сократится.
s = fr"""
Тогда
$$
n = \frac{{{t}^2 \sigma^2}}{{\left(\frac{{\sigma}}{denom}\right)^2}} = {t}^2 \cdot {round(denom**2)} = {round(t**2 * denom**2, 2)}  \approx \textbf{{{n_min}}}.
$$

Таким образом, минимальный объем выборки, позволяющий с доверительной вероятностью $\gamma = {gamma}$ утверждать, что, принимая среднее арифметическое за математическое ожидание случайной величины $X$, допуская погрешность $\varepsilon = \frac{{\sigma}}{{2}}$, равен $\textbf{{{n_min}}}$.
"""
display(Markdown(s))
```
