####Решение

Составим вспомогательную таблицу

```python
# @title
# Используем xi, freq, nsize, nnn из предыдущей ячейки
import math

# Вычисляем выборочную среднюю
xB = sum(xi * freq) / nsize

# Вычисляем сумму n_i * x_i^2
sum_ni_xi_sq = sum(freq * xi**2)

# Выборочная дисперсия (смещенная оценка)
sigmaB = round(math.sqrt(sum_ni_xi_sq / nsize - xB**2), 1)

# Шаг между вариантами
h = xi[1] - xi[0]

# Коэффициент для теоретических частот
coef = nsize * h / sigmaB

# Вычисляем u_i, φ(u_i), теоретические частоты
ui = (xi - xB) / sigmaB
phi_list = [ (1/math.sqrt(2*math.pi)) * math.exp(-u**2/2) for u in ui ]
n_prime = [coef * p for p in phi_list]

# Округляем теоретические частоты до целых
n_prime_int = [round(x) for x in n_prime]

# Корректируем сумму, чтобы была равна nsize
diff = nsize - sum(n_prime_int)
if diff != 0:
    # Добавляем разницу к наибольшей частоте
    idx_max = n_prime_int.index(max(n_prime_int))
    n_prime_int[idx_max] += diff


xini_sum = sum(xi*freq)

x2ni_sum = sum(freq * xi**2)

# Формируем таблицу
headers = [""] + [str(" ") for i in range(nnn)]
aux = [
    ["$x_i$"] + [f"{x:.2f}".rstrip('0').rstrip('.') for x in xi] + [fr"$\sum$"],
    ["$n_i$"] + [str(int(n)) for n in freq] + [fr"${nsize}$"],
    ["$x_i n_i$"] + [f"{x:.2f}".rstrip('0').rstrip('.') for x in xi*freq] + [fr"${round(xini_sum, 1)}$"],
    ["$x_i - \\bar{x}_B$"] + [f"{x:.2f}".rstrip('0').rstrip('.') for x in xi - xB] + [" "],
    ["$n_i x_i^2$"] + [f"{x:.2f}".rstrip('0').rstrip('.') for x in freq * xi**2] + [fr"${round(x2ni_sum, 2)}$"],
    ["$u_i$"] + [f"{x:.2f}".rstrip('0').rstrip('.') for x in ui] + [" "],
    ["$\\varphi(u_i)$"] + [f"{x:.4f}".rstrip('0').rstrip('.') for x in phi_list] + [" "],
    ["$n_i'$"] + [str(int(x)) for x in n_prime_int] + [fr"${sum(n_prime_int)}$"]
]
rows = []

for i in range (len(aux[0])):
  rows.append([
      aux[0][i],
      aux[1][i],
      aux[2][i],
      aux[3][i],
      aux[4][i],
      aux[5][i],
      aux[6][i],
      aux[7][i],
  ])

h2 = [str(i+1) for i in range(len(rows[0]))] + [" ", " "]
h3 = ["$x_i$", "$n_i$", "$x_i n_i$", "$x_i - \\bar{x}_B$", "$n_i x_i^2$","$u_i$", "$\\varphi(u_i)$", "$n_i'$", " ", " "]
# Собираем Markdown таблицу
table_md = "| " + " | ".join(headers) + " |\n"
table_md = "| " + " | ".join(h2) + " |\n"
#table_md = "| " + " | ".join(h3) + " |\n"
table_md += "|" + "|".join(["---"] * len(headers)) + "|\n"
for row in rows:
    table_md += "| " + " | ".join(row) + " |\n"

# Выводим заголовок и таблицу
display(Markdown("####Таблица 1 – Расчет теоретических частот"))
display(Markdown(table_md))
```

```python
# @title
xini_sum_fmt = f"{xini_sum:.2f}".rstrip('0').rstrip('.')
nsize_fmt = str(int(nsize))
xB_fmt = round(xB, 1)

s = fr"""
По результату колонки 3 находим:

$$
\bar{{x}}_B = \frac{{\sum x_i n_i}}{{n}} = \frac{{{xini_sum_fmt}}}{{{nsize_fmt}}} = \textbf{{{xB_fmt}}}.
$$
"""
display(Markdown(s))
```
