```python
# @title
# Исходные данные уже загружены: x1, y1, Dx, Dy
def format_n_nonzero_decimals(num, nz):

    integer_part = int(math.trunc(num))
    decimal_part = abs(num) - abs(integer_part)

    if decimal_part == 0:
        return f"{integer_part}"


    first_nonzero_pos = math.ceil(-math.log10(decimal_part))

    # We need 3 digits starting from the first non-zero position
    total_decimals_needed = first_nonzero_pos + nz -1

    # Round the original number to this specific depth and format as string
    return f"{round(num, total_decimals_needed)}"




x1_mean = np.mean(x1)
y1_mean = np.mean(y1)
sum_x1 = np.sum(x1)
sum_y1 = np.sum(y1)
nx = len(x1)
ny = len(y1)




x1_mean_fmt = f"{format_n_nonzero_decimals(x1_mean, 3)}"
y1_mean_fmt = f"{format_n_nonzero_decimals(y1_mean, 3)}"
s = fr"""
**1.** а) Вычислим выборочные средние:

$$
\bar{{x}}_B = \frac{{\sum x_i}}{{n_x}} = \frac{{{sum_x1:.1f}}}{{{nx}}} = \textbf{{ {x1_mean_fmt} }};
$$

$$
\bar{{y}}_B = \frac{{\sum y_i}}{{n_y}} = \frac{{{sum_y1:.1f}}}{{{ny}}} = \textbf {{ {y1_mean_fmt} }}.
$$
"""
display(Markdown(s))
```

```python
# @title
import math

Dx_val = Dx
Dy_val = Dy

Z_obs = (x1_mean - y1_mean) / math.sqrt(Dx_val/nx + Dy_val/ny)
Z_obs_fmt = f"{Z_obs:.2f}".rstrip('0').rstrip('.')

s = fr"""
Так как дисперсии генеральных совокупностей известны, наблюдаемое значение критерия найдём по формуле:

$$
Z_{{\text{{}}}} = \frac{{\bar{{x}}_B - \bar{{y}}_B}}{{\sqrt{{\frac{{D_x}}{{n_x}} + \frac{{D_y}}{{n_y}}}}}} =
\frac{{{x1_mean_fmt} - {y1_mean_fmt}}}{{\sqrt{{\frac{{{Dx_val}}}{{{nx}}} + \frac{{{Dy_val}}}{{{ny}}}}}}} = \textbf{{ {Z_obs_fmt} }}.
$$
"""
display(Markdown(s))
```

Поэтому проверим сначала гипотезу о равенстве дисперсий генеральных совокупностей, используя критерий Фишера-Снедекора.


```python
# @title




s_x2 = np.var(x1, ddof=1)
s_y2 = np.var(y1, ddof=1)

s_x2_fmt = f"{format_n_nonzero_decimals(s_x2, 2)}"
s_y2_fmt = f"{format_n_nonzero_decimals(s_y2, 2)}"

s = fr"""
Вычислим исправленные выборочные дисперсии:

$$
s_x^2 = \frac{{1}}{{n_x-1}} \sum (x_i - \bar{{x}}_B)^2 = \textbf{{{s_x2_fmt}}} - \text{{исправленная дисперсия первой выборки}}
$$

$$
s_y^2 = \frac{{1}}{{n_y-1}} \sum (y_i - \bar{{y}}_B)^2 = \textbf{{{s_y2_fmt}}} - \text{{исправленная дисперсия второй выборки}}
$$
"""
display(Markdown(s))
```

```python
# @title


k1_str = fr"n_x"
k2_str = fr"n_y"

# Большую дисперсию помещаем в числитель
if s_x2 >= s_y2:
    F_obs = s_x2 / s_y2
    df_big = nx - 1
    df_small = ny - 1
    numerator_var = 's_x^2'
    denominator_var = 's_y^2'
else:
    F_obs = s_y2 / s_x2
    df_big = ny - 1
    df_small = nx - 1
    numerator_var = 's_y^2'
    denominator_var = 's_x^2'
    k1_str = fr"n_y"
    k2_str = fr"n_x"

F_obs_fmt = f"{F_obs:.2f}".rstrip('0').rstrip('.')

s = fr"""
Так как {('первая дисперсия больше', 'вторая дисперсия больше')[numerator_var == 's_y^2']},
помещаем её в числитель формулы:

$$
F = \frac{{{numerator_var}}}{{{denominator_var}}} =
\frac{{{s_x2_fmt if numerator_var == 's_x^2' else s_y2_fmt}}}
{{{s_y2_fmt if numerator_var == 's_x^2' else s_x2_fmt}}} = \textbf{{ {F_obs_fmt} }}.
$$
"""
display(Markdown(s))
```

```python
# @title


F_crit = stats.f.ppf(1 - alpha, df_big, df_small)
F_crit_fmt = f"{F_crit:.2f}".rstrip('0').rstrip('.')

s = fr"""
Критическое значение $F_{{\text{{крит}}}}$ найдём с помощью таблицы распределения
Фишера-Снедекора (приложение 7) по уровню значимости $\alpha = 0.05$ и числам степеней свободы
$k_1 = {k1_str} - 1 = {df_big + 1} - 1 = {df_big}$  и $k_2 = {k2_str} - 1 = {df_small + 1} - 1 = {df_small}$ находим:

$$
F_{{\text{{крит}}}} = F({alpha}; {df_big}; {df_small}) = \textbf{{ {F_crit_fmt} }}.
$$
"""
display(Markdown(s))
```

*[$k_1$ $-$ число степеней свободы большей по величине дисперсии; $k_2$ $-$ число степеней свободы меньшей по величине дисперсии]*

```python
# @title

# Данные уже загружены: x2_vals, x2_freq, y2_vals, y2_freq
# Вычисляем объёмы выборок
if (variant != 0):
  n_x2 = int(np.sum(x2_freq))
  n_y2 = int(np.sum(y2_freq))
else:
  n_x2 = 100
  n_y2 = 120

# Выборочные средние
if(variant != 0):
  x_mean2 = np.sum(x2_vals * x2_freq) / n_x2
  y_mean2 = np.sum(y2_vals * y2_freq) / n_y2
else:
  x_mean2 = 32.4
  y_mean2 = 30.1

#  выборочные несмещённые дисперсии
if (variant != 0):
  s2_x2 = np.sum(x2_freq * (x2_vals - x_mean2)**2) / (n_x2 - 1)
  s2_y2 = np.sum(y2_freq * (y2_vals - y_mean2)**2) / (n_y2 - 1)
else:
  s2_x2 = 15
  s2_y2 = 25

# Наблюдаемое значение приближенной нормальной статистики
Z_obs2 = (x_mean2 - y_mean2) / np.sqrt(s2_x2/n_x2 + s2_y2/n_y2)

# Критическое значение для левосторонней критической области
alpha = 0.05
Z_crit2 = stats.norm.ppf(alpha)   # левосторонний квантиль

# Форматирование для вывода
x_mean2_fmt = f"{x_mean2:.4f}".rstrip('0').rstrip('.')
y_mean2_fmt = f"{y_mean2:.4f}".rstrip('0').rstrip('.')
s2_x2_fmt = f"{s2_x2:.4f}".rstrip('0').rstrip('.')
s2_y2_fmt = f"{s2_y2:.4f}".rstrip('0').rstrip('.')
Z_obs2_fmt = f"{Z_obs2:.4f}".rstrip('0').rstrip('.')
Z_crit2_fmt = f"{Z_crit2:.4f}".rstrip('0').rstrip('.')

# Вывод решения
s = fr"""
Воспользуемся приближенно нормально распределённой величиной $Z'$.
Вычислим выборочные характеристики по данным второй части задания.

Объёмы выборок: $n_x = {n_x2}$, $n_y = {n_y2}$.

Выборочные средние:

$$
\bar{{x}}_B = \frac{{\sum x_i n_i}}{{n_x}} = {x_mean2_fmt},
\qquad
\bar{{y}}_B = \frac{{\sum y_i n_i}}{{n_y}} = {y_mean2_fmt}.
$$

Выборочные несмещённые дисперсии:

$$
s_x^2 = \frac{{\sum (x_i - \bar{{x}}_B)^2 n_i}}{{n_x - 1}} = {s2_x2_fmt},
\qquad
s_y^2 = \frac{{\sum (y_i - \bar{{y}}_B)^2 n_i}}{{n_y - 1}} = {s2_y2_fmt}.
$$

Наблюдаемое значение статистики:

$$
Z'_{{\text{{набл}}}} = \frac{{\bar{{x}}_B - \bar{{y}}_B}}{{\sqrt{{\frac{{(D_x)_B}}{{n_x}} + \frac{{(D_y)_B}}{{n_y}}}}}} =
\frac{{{x_mean2_fmt} - {y_mean2_fmt}}}{{\sqrt{{\frac{{{s2_x2_fmt}}}{{{n_x2}}} + \frac{{{s2_y2_fmt}}}{{{n_y2}}}}}}} = \textbf{{ {Z_obs2_fmt} }}.
$$

При конкурирующей гипотезе $H_1: M_x < M_y$ критическая область левосторонняя.
Критическое значение находим из условия:

$$
\Phi(Z'_{{\text{{крит}}}}) =  0.05.
$$


Используя обратную функцию Лапласа, находим $Z'_{{\text{{крит}}}} = \textbf{{ {Z_crit2_fmt} }}$.

Сравниваем:

$$
Z'_{{\text{{набл}}}} = {Z_obs2_fmt} \quad \text{{и}} \quad Z'_{{\text{{крит}}}} = {Z_crit2_fmt}.
$$
"""
display(Markdown(s))

# Логический вывод
sg = "<"
conclusion = ""


if (Z_crit2 < 0):
  Zstr = "Z_{набл}'"
  if Z_obs2 <= Z_crit2:
    conclusion = "отвергаем нулевую гипотезу о равенстве математических ожиданий"
  else:
    conclusion = "нет оснований отвергнуть нулевую гипотезу о равенстве математических ожиданий"
    sg = fr"\geq"
else:
  if abs(Z_obs2) >= Z_crit2:
      conclusion = "отвергаем нулевую гипотезу о равенстве математических ожиданий"
      sg = fr"\geq"
  else:
      conclusion = "нет оснований отвергнуть нулевую гипотезу о равенстве математических ожиданий"




s2 = f"""
Так как $ Z_{{набл}} {sg} Z'_{{крит}}$, то **{conclusion}** .
"""
display(Markdown(s2))
```
