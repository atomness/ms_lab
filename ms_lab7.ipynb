2) в предположении, что между $x$ и $y$ существует квадратичная зависимость, определить ее эмпирическое уравнение;
3) найти сумму квадратов отклонений для найденных зависимостей сравнить качество приближений."

#### Решение

Составим вспомогательную таблицу

```python
# @title



n = len(X)

# Суммы для нормальной системы
sum_x = np.sum(X)
sum_y = np.sum(Y)
sum_x2 = np.sum(X**2)
sum_x3 = np.sum(X**3)
sum_x4 = np.sum(X**4)
sum_xy = np.sum(X * Y)
sum_x2y = np.sum(X**2 * Y)

def fmt_num(val, precision=2):
    """Форматирование чисел: целые без .0, дробные с первыми precision значащими цифрами после запятой"""

    integer_part = int(math.trunc(val))
    decimal_part = abs(val) - abs(integer_part)

    if decimal_part < 1e-9:
        return f"{int(integer_part)}"


    first_nonzero_pos = math.ceil(-math.log10(decimal_part))

    # We need 3 digits starting from the first non-zero position
    total_decimals_needed = first_nonzero_pos + precision -1

    # Round the original number to this specific depth and format as string
    return f"{round(val, total_decimals_needed)}"

table_title = "Таблица 1 – Расчеты сумм нормальной системы уравнений"
header = ["№", "$x_i$", "$y_i$", "$x_i^2$", "$x_i^3$", "$x_i^4$", "$x_i y_i$", "$x_i^2 y_i$"]

rows = []
for i in range(n):
    xi = X[i]
    yi = Y[i]
    rows.append([
        str(i + 1),
        fmt_num(xi),
        fmt_num(yi),
        fmt_num(xi**2),
        fmt_num(xi**3),
        fmt_num(xi**4),
        fmt_num(xi * yi),
        fmt_num(xi**2 * yi)
    ])

sum_row = [fr"$\sum$",
           fmt_num(sum_x),
           fmt_num(sum_y),
           fmt_num(sum_x2),
           fmt_num(sum_x3),
           fmt_num(sum_x4),
           fmt_num(sum_xy),
           fmt_num(sum_x2y)]
rows.append(sum_row)

table_md = f" **{table_title}** \n\n"
table_md += "| " + " | ".join(header) + " |\n"
table_md += "|" + "|".join(["---"] * len(header)) + "|\n"
for row in rows:
    table_md += "| " + " | ".join(row) + " |\n"

display(Markdown(table_md))
```

```python
# Найдем уравнение линейной зависимости
from IPython.display import display, Markdown

text = f"""
Найдем уравнение линейной зависимости. Для этого составим систему уравнений

$$
\\begin{{cases}}
   a \\cdot {fmt_num(sum_x2)} + b \\cdot {fmt_num(sum_x)} = {fmt_num(sum_xy)}, \\\\
   a \\cdot {fmt_num(sum_x)} + b \\cdot {n}  = {fmt_num(sum_y)}.
\\end{{cases}}
$$
"""
display(Markdown(text))
```

```python
# @title
# Решим систему по формулам Крамера
from IPython.display import display, Markdown

# Определители
Delta = n * sum_x2 - sum_x**2
Delta_a = n * sum_xy - sum_x * sum_y
Delta_b = sum_y * sum_x2 - sum_x * sum_xy

a_val = Delta_a / Delta
b_val = Delta_b / Delta

# Форматирование коэффициентов с округлением до 4 знаков, без лишних нулей


a_str = fmt_num(a_val)
b_str = fmt_num(b_val)

# Построение уравнения с учётом знака b
if b_val < 0:
    equation = f"y = {a_str}x - {abs(b_val):.4f}".rstrip('0').rstrip('.')
else:
    equation = f"y = {a_str}x + {b_str}"

text = f"""
Решим систему по формулам Крамера:


$$
\\Delta = {n} \\cdot {fmt_num(sum_x2)} - ({fmt_num(sum_x)})^2 = {fmt_num(Delta)}
$$

$$
\\Delta_a = {n} \\cdot {fmt_num(sum_xy)} - {fmt_num(sum_x)} \\cdot {fmt_num(sum_y)} = {fmt_num(Delta_a)}
$$

$$
\\Delta_b = {fmt_num(sum_y)} \\cdot {fmt_num(sum_x2)} - {fmt_num(sum_x)} \\cdot {fmt_num(sum_xy)} = {fmt_num(Delta_b)}
$$

Отсюда

$$
a = \\frac{{\\Delta_a}}{{\\Delta}} = \\frac{{{fmt_num(Delta_a)}}}{{{fmt_num(Delta)}}} \\approx {a_str}
$$

$$
b = \\frac{{\\Delta_b}}{{\\Delta}} = \\frac{{{fmt_num(Delta_b)}}}{{{fmt_num(Delta)}}} \\approx {b_str}
$$

${equation}$ – искомое эмпирическое уравнение линейной зависимости.
"""
display(Markdown(text))
```

```python
# @title
# Далее найдем уравнение квадратичной зависимости
from IPython.display import display, Markdown

text = f"""
Далее найдем уравнение квадратичной зависимости. Нормальная система уравнений примет вид:



$$
\\begin{{cases}}
   a \\cdot {fmt_num(sum_x4)} + b \\cdot {fmt_num(sum_x3)} + c \\cdot {fmt_num(sum_x2)}  = {fmt_num(sum_x2y)}, \\\\
    a \\cdot {fmt_num(sum_x3)} + b \\cdot {fmt_num(sum_x2)} + c \\cdot {fmt_num(sum_x)} = {fmt_num(sum_xy)}, \\\\
   a \\cdot {fmt_num(sum_x2)}  + b \\cdot {fmt_num(sum_x)}  + c \\cdot {n}= {fmt_num(sum_y)}.
\\end{{cases}}
$$
"""
display(Markdown(text))
```

```python
# @title
# Решая систему получим
from IPython.display import display, Markdown
import numpy as np

# Решение системы [c, b, a]
A = np.array([
    [n, sum_x, sum_x2],
    [sum_x, sum_x2, sum_x3],
    [sum_x2, sum_x3, sum_x4]
])
rhs = np.array([sum_y, sum_xy, sum_x2y])
c_quad, b_quad, a_quad = np.linalg.solve(A, rhs)

# Форматирование
def fmt_coef(val):
    s = f"{val:.4f}".rstrip('0').rstrip('.')
    if s == '-0':
        s = '0'
    return s

a_q_str = fmt_coef(a_quad)
b_q_str = fmt_coef(b_quad)
c_q_str = fmt_coef(c_quad)

# Уравнение
if b_quad < 0:
    eq = f"y = {a_q_str}x^2 - {fmt_coef(abs(b_quad))}x + {c_q_str}"
elif b_quad > 0:
    eq = f"y = {a_q_str}x^2 + {b_q_str}x + {c_q_str}"
else:
    eq = f"y = {a_q_str}x^2 + {c_q_str}"

text = f"""
Решая систему получим: $a = {a_q_str}$, $b = {b_q_str}$, $c = {c_q_str}$.

${eq}$ $-$ искомое эмпирическое уравнение квадратичной зависимости.
"""
display(Markdown(text))
```

Найдем сумму квадратов отклонений для найденных зависимостей. Вычисления оформим в таблицу

```python
# @title
# Найдем сумму квадратов отклонений
from IPython.display import display, Markdown
import numpy as np

# Предсказанные значения
y_lin_pred = a_val * X + b_val
y_quad_pred = a_quad * X**2 + b_quad * X + c_quad

# Отклонения
eps_lin = y_lin_pred - Y
eps_quad = y_quad_pred - Y

# Квадраты отклонений
eps_lin_sq = eps_lin**2
eps_quad_sq = eps_quad**2

# Суммы квадратов
S_lin = np.sum(eps_lin_sq)
S_quad = np.sum(eps_quad_sq)

# Формирование заголовков уравнений
def format_lin_eq(a, b):
    if b < 0:
        return f"{fmt_coef(a)}x_i - {fmt_coef(abs(b))}"
    elif b > 0:
        return f"{fmt_coef(a)}x_i + {fmt_coef(b)}"
    else:
        return f"{fmt_coef(a)}x_i"

def format_quad_eq(a, b, c):
    eq = f"{fmt_coef(a)}x_i^2"
    if b < 0:
        eq += f" - {fmt_coef(abs(b))}x_i"
    elif b > 0:
        eq += f" + {fmt_coef(b)}x_i"
    if c < 0:
        eq += f" - {fmt_coef(abs(c))}"
    elif c > 0:
        eq += f" + {fmt_coef(c)}"
    return eq

lin_eq_str = format_lin_eq(a_val, b_val)
quad_eq_str = format_quad_eq(a_quad, b_quad, c_quad)

header = ["$x_i$", "$y_i$", f"${lin_eq_str}$", "$\\varepsilon_i$", "$(\\varepsilon_i)^2$", f"${quad_eq_str}$", "$\\varepsilon'_i$", "$(\\varepsilon'_i)^2$"]

rows = []
for i in range(n):
    xi = X[i]
    yi = Y[i]
    lin_pred = y_lin_pred[i]
    quad_pred = y_quad_pred[i]
    e_lin = eps_lin[i]
    e_quad = eps_quad[i]
    e_lin_sq = eps_lin_sq[i]
    e_quad_sq = eps_quad_sq[i]
    rows.append([
        fmt_num(xi),
        fmt_num(yi),
        fmt_num(lin_pred),
        fmt_num(e_lin),
        fmt_num(e_lin_sq),
        fmt_num(quad_pred),
        fmt_num(e_quad),
        fmt_num(e_quad_sq)
    ])

# Добавим строку сумм для квадратов отклонений? В примере в таблице есть строка сумм, но в столбцах для сумм у них пусто для xi, yi, y_lin_pred, y_quad_pred? В примере после последней строки данных они пишут "Σ" и только сумма квадратов отклонений (2.6816 и 134.242). Они не суммируют остальные столбцы. Мы можем добавить строку с суммами только для квадратов отклонений, а в остальных ячейках поставить прочерк или пусто. Лучше сделать как в примере: в строке сумм в первом столбце "Σ", затем в столбцах xi, yi, y_lin_pred, y_quad_pred, ε_i, ε'_i пусто, а в столбцах (ε_i)^2 и (ε'_i)^2 поставить суммы. В примере таблица 10 имеет после данных строку "Σ" с суммой квадратов отклонений для каждой модели (в колонках (εi)^2 и (ε'i)^2). Они не выводили суммы для остальных величин. Мы можем так сделать.

# Строка сумм
sum_row = [fr"$\sum$"] + [""] * 7  # 8 столбцов, заполним позже
sum_row[4] = fmt_num(S_lin)   # индекс колонки (ε_i)^2 — это 5-я (0-based: 0 xi,1 yi,2 lin,3 eps,4 eps_sq,5 quad,6 eps_quad,7 eps_quad_sq)
sum_row[7] = fmt_num(S_quad)  # индекс колонки (ε'_i)^2 — 7
rows.append(sum_row)

table_md = "**Таблица 10 – Вычисление квадратов отклонений**\n\n"
table_md += "| " + " | ".join(header) + " |\n"
table_md += "|" + "|".join(["---"] * len(header)) + "|\n"
for row in rows:
    table_md += "| " + " | ".join(row) + " |\n"

display(Markdown(table_md))
```

```python
# @title
# Вывод о качестве приближения
from IPython.display import display, Markdown

if S_lin < S_quad:
    conclusion = ("Как видим из таблицы, сумма квадратов отклонений для квадратичной зависимости "
                  "многократно превышает сумму квадратов отклонений для линейной зависимости. "
                  "Таким образом, линейная зависимость дает более качественное приближение, чем квадратичная.")
else:
    conclusion = ("Как видим из таблицы, сумма квадратов отклонений для линейной зависимости "
                  "превышает сумму квадратов отклонений для квадратичной зависимости. "
                  "Таким образом, квадратичная зависимость дает более качественное приближение, чем линейная.")

display(Markdown(conclusion))
```
